## 7.2.2 Using Functions or Mappings for Data Transformation

```python
data = pd.DataFrame({'food': ['bacon', 'pulled pork', 'bacon',
                              'pastrami', 'corned beef', 'bacon',
                              'pastrami', 'honey ham', 'nova lox'],
                     'ounces': [4, 3, 12, 6, 7.5, 8, 3, 5, 6]})
```

```python
data
```

```python
meat_to_animal = {
    'bacon': 'pig',
    'pulled pork': 'pig',
    'pastrami': 'cow',
    'corned beef': 'cow',
    'honey ham': 'pig',
    'nova lox': 'salmon'
}
```

```python
data['animal'] = data['food'].map(meat_to_animal)
```

```python
data
```

```python
def get_animal(x):
    return meat_to_animal[x]
```

```python
data['food'].map(get_animal)
```

## 7.2.4 Rename Axis Index

```python
data = pd.DataFrame(np.arange(12).reshape((3, 4)),
                    index=['Ohio', 'Colorado', 'New York'],
                    columns=['one', 'two', 'three', 'four'])
```

```python
def transform(x):
    return x[:4].upper()
```

```python
data.index.map(transform)
```

```python
data.index = data.index.map(transform)
```

```python
data
```

```python
data.rename(index=str.title, columns=str.upper)
```

```python
data.rename(index={'OHIO': 'INDIANA'},
            columns={'three': 'peekaboo'})
```

## 7.2.5 Discretization and Binning

```python
ages = [20, 22, 25, 27, 21, 23, 37, 31, 61, 45, 41, 32]
```

```python
bins = [18, 25, 35, 60, 100]
```

```python
age_categories = pd.cut(ages, bins)
```

```python
age_categories
```

```python
age_categories.codes
```

```python
age_categories.categories
```

```python
age_categories.categories[0]
```

```python
age_categories.value_counts(age_categories)
```

```python
pd.cut(ages, bins, right=False)
```

```python
group_names = ['Youth', 'YoungAdult', 'MiddleAged', 'Senior']
```

```python
pd.cut(ages, bins, labels=group_names)
```

```python
data = np.random.uniform(size=10)
```

```python
pd.cut(data, 4, precision=2)
```

```python
data = np.random.standard_normal(1000)
```

```python
quartiles = pd.qcut(data, 4, precision=2)
```

```python
quartiles
```

```python
quartiles.value_counts(quartiles)
```

```python
pd.qcut(data, [0, 0.1, 0.5, 0.9, 1.]).value_counts()
```

## 7.2.6 Detecting and Filtering Outliers

```python
data = pd.DataFrame(np.random.standard_normal((1000, 4)))
```

```python
data.describe()
```

```python
col = data[2]
```

```python
col[col.abs() > 3]
```

```python
data[(data.abs() > 3).any(axis='columns')]
```

```python
data[data.abs() > 3] = np.sign(data) * 3
```

```python
data.describe()
```

```python
np.sign(data).head()
```

## 7.2.7 Replacement and Random Sampling

```python
df = pd.DataFrame(np.arange(5 * 7).reshape((5, 7)))
```

```python
df
```

```python
sampler = np.random.permutation(5)
```

```python
sampler
```

```python
df.take(sampler)
```

```python
df.iloc[sampler]
```

```python
column_sampler = np.random.permutation(7)
```

```python
column_sampler
```

```python
df.take(column_sampler, axis='columns')
```

```python
df.sample(n=3)
```

```python
choices = pd.Series([5, 7, -1, 6, 4])
```

```python
choices.sample(n=10, replace=True)
```
