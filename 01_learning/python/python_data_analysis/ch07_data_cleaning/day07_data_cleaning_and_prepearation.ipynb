## 7.1.1 Filter Out Missing Data

```python
data = pd.Series([1, np.nan, 3.5, np.nan, 7])
```

```python
data.dropna()
```

```python
data[data.notna()]
```

```python
data = pd.DataFrame([[1., 6.5, 3.], [1., np.nan, np.nan],
                     [np.nan, np.nan, np.nan], [np.nan, 6.5, 3.]])
```

```python
data
```

```python
data.dropna()
```

```python
data.dropna(how='all')
```

```python
data[4] = np.nan
```

```python
data
```

```python
data.dropna(axis='columns', how='all')
```

```python
df = pd.DataFrame(np.random.standard_normal((7,3)))
```

```python
df.iloc[:4, 1] = np.nan
```

```python
df.iloc[:2, 2] = np.nan
```

```python
df
```

```python
df.dropna()
```

```python
df.dropna(thresh=2)
```

## 7.1.2 Imputing Missing Data

```python
df.fillna(0)
```

```python
df.fillna({1: 0.5, 2: 0})
```

```python
df = pd.DataFrame(np.random.standard_normal((6, 3)))
```

```python
df.iloc[2:, 1] = np.nan
```

```python
df.iloc[4:, 2] = np.nan
```

```python
df
```

```python
df.ffill()
```

```python
df.ffill(limit = 2)
```

```python
data = pd.Series([1., np.nan, 3.5, np.nan, 7])
```

```python
data.fillna(data.mean())
```
