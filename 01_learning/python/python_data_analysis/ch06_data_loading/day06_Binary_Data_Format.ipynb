## 6.2.2 Using the HDF5 format

```python
conda install pytables
```

```python
frame = pd.DataFrame({'a': np.random.standard_normal(100)}) 
```

```python
store = pd.HDFStore('pydata-book/examples/mydata.h5')
```

```python
store['obj1'] = frame
```

```python
store['obj1_col'] = frame['a']
```

```python
store
```

```python
store['obj1']
```

```python
store.put('obj2', frame, format='table')
```

```python
store.select('obj2', where=['index >= 10 and index <= 15'])
```

```python
store.close
```

```python
frame.to_hdf('pydata-book/examples/mydata.h5', 'obj3', format='table')
```

```python
pd.read_hdf('pydata-book/examples/mydata.h5', 'obj3', where=['index < 5'])
```
