# Summary and Calculation of Descriptive Statistics

```python
import pandas as pd
import numpy as np
```

```python
df = pd.DataFrame([[1.4, np.nan], [7.1, -4.5],
                   [np.nan, np.nan], [0.75, -1.3]],
                   index=['a', 'b', 'c', 'd'],
                   columns=['one', 'two']) 
```

```python
df
```

```python
df.sum() 
```

```python
df.sum(axis='columns') 
```

```python
df.sum(axis='index', skipna=False)
```

```python
df.sum(axis='columns', skipna=False)
```

```python
df.mean(axis='columns')
```

```python
df.idxmax()
```

```python
df.cumsum()
```

```python
df.describe()
```
