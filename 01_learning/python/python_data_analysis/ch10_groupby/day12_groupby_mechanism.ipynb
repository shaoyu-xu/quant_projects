# 10.1 Groupby Mechanism

```python
import pandas as pd
import numpy as np
```

```python
df = pd.DataFrame({'key1' : ['a', 'a', None, 'b', 'b', 'a', None],
                   'key2' : pd.Series([1, 2, 1, 2, 1, None, 1], dtype='Int64'),
                   'data1' : np.random.standard_normal(7),
                   'data2' : np.random.standard_normal(7)
                   })
```

```python
df
```

```python
grouped = df['data1'].groupby(df['key1']) # groupby() returns a GroupBy object, which is an intermediate object that contains information about the groups but does not yet perform any computations. You can think of it as a "recipe" for how to group the data.
```

```python
grouped
```

```python
grouped.mean() # The mean() method is called on the GroupBy object, which triggers the actual computation of the mean for each group. The result is a new Series with the group labels as the index and the computed means as the values.
```

```python
means = df['data1'].groupby([df['key1'], df['key2']]).mean() # You can also group by multiple keys by passing a list of keys to the groupby() method. The result is a Series with a MultiIndex, where each level of the index corresponds to one of the grouping keys.
```

```python
means
```

```python
means.unstack() # The unstack() method is called on the Series with a MultiIndex, which reshapes the data into a DataFrame where one of the levels of the index becomes the columns. This makes it easier to visualize and analyze the grouped data.
```

```python
states = np.array(['OH', 'CA', 'CA', 'OH', 'OH', 'CA', 'OH']) # You can also group by arrays of values, such as a NumPy array. The result is a Series with the array values as the index and the computed means as the values.
```

```python
years = [2005, 2005, 2006, 2005, 2006, 2005, 2006]
```

```python
df['data1'].groupby([states, years]).mean() # You can also group by multiple arrays of values, such as a NumPy array and a list. The result is a Series with a MultiIndex, where each level of the index corresponds to one of the grouping keys.
```

```python
df.groupby('key1').mean() # You can also group by a single key and compute the mean for all numeric columns in the DataFrame. The result is a new DataFrame with the group labels as the index and the computed means for each numeric column as the values.
```

```python
df.groupby('key2')[['data1', 'data2']].mean() # You can also group by a single key and compute the mean for all numeric columns in the DataFrame. The result is a new DataFrame with the group labels as the index and the computed means for each numeric column as the values.
```

```python
df.groupby(['key1', 'key2']).mean() # You can also group by multiple keys and compute the mean for all numeric columns in the DataFrame. The result is a new DataFrame with a MultiIndex, where each level of the index corresponds to one of the grouping keys, and the computed means for each numeric column as the values.
```

```python
df.groupby(['key1', 'key2']).size() # The size() method is called on the GroupBy object, which computes the size of each group. The result is a Series with a MultiIndex, where each level of the index corresponds to one of the grouping keys, and the computed sizes for each group as the values.
```

```python
df.groupby('key1', dropna=False).size() # The dropna parameter is set to False, which means that groups with NaN values in the grouping key will be included in the result. The result is a Series with the group labels as the index and the computed sizes for each group as the values, including a group for NaN values.
```

```python
df.groupby(['key1', 'key2'], dropna=False).size()
```

```python
df.groupby('key1').count() # The count() method is called on the GroupBy object, 
# which computes the number of non-NA/null observations in each group. The result is a new DataFrame with the group labels as the index and the computed counts for each column as the values.
```

## 10.1.1 Iterate over the groups

```python
for name, group in df.groupby('key1'):
    print(name)
    print(group)
# The groupby() method can also be used in a for loop to iterate over the groups.
# The result is a sequence of (group name, group DataFrame) pairs, where each group name corresponds to a unique value in the grouping key and each group DataFrame contains the rows of the original DataFrame that belong to that group.
```

```python
for (k1, k2), group in df.groupby(['key1', 'key2']):
    print((k1, k2))
    print(group)
# You can also group by multiple keys and iterate over the groups. The result is a sequence
```

```python
pieces = {name: group for name, group in df.groupby('key1')}
# The groupby() method can also be used in a dictionary comprehension to create a dictionary of DataFrames, where each key is a unique value in the grouping key and each value is the corresponding group DataFrame.
```

```python
pieces
```

```python
pieces['b']
```

```python
grouped = df.T.groupby({'key1': 'key', 'key2': 'key',
                     'data1': 'data', 'data2': 'data'}) # You can also group by a mapping of column names to group names, which allows you to group columns instead of rows. The result is a GroupBy object that groups the columns based on the specified mapping.
```

```python
for group_key, group_values in grouped:
    print(group_key)
    print(group_values)
```

```python
grouped
```

## 10.1.2 Select one or more columns

```python
df.groupby(['key1', 'key2'])[['data2']].mean() 
# The groupby() method can also be used to group by multiple keys and select specific columns for aggregation. The result is a new DataFrame with a MultiIndex, where each level of the index corresponds to one of the grouping keys, and the computed means for the selected column as the values.
```

```python
df
```

```python
s_grouped = df.groupby(['key1', 'key2'])['data2']
```

## 10.1.3 Grouping Using Dictionaries and Series

```python
people = pd.DataFrame(np.random.standard_normal((5, 5)),
                      columns=['a', 'b', 'c', 'd', 'e'],
                      index=['Joe', 'Steve', 'Wanda', 'Jill', 'Trey'])
```

```python
people.iloc[2:3, [1, 2]] = np.nan # Add some NaN values to the DataFrame to demonstrate the effect of the dropna parameter in groupby operations.
```

```python
people
```

```python
mapping = {'a': 'red', 'b': 'red', 'c': 'blue', 'd': 'blue', 'e': 'red', 'f' : 'orange'}
```

```python
by_column = people.T.groupby(mapping)
```

```python
by_column.sum().T
```

```python
map_series = pd.Series(mapping)
```

```python
map_series
```

```python
people.T.groupby(map_series).count().T
```

## 10.1.4 Grouping Using Functions

```python
people.groupby(len).sum() # You can also group by the length of the index labels, which allows you to group rows based on the length of their index labels. The result is a new DataFrame with the computed sums for each group as the values.
```

```python
key_list = ['one', 'one', 'one', 'two', 'two']
```

```python
people.groupby([len, key_list]).min()
```

## 10.1.5 Grouping by Index Levels

```python
columns = pd.MultiIndex.from_arrays([['US', 'US', 'US', 'JP', 'JP'],
                                     [1, 3, 5, 1, 3]],
                                     names=['cty', 'tenor'])
```

```python
hier_df = pd.DataFrame(np.random.standard_normal((4, 5)), columns=columns)
```

```python
hier_df
```

```python
hier_df.T.groupby(level='cty').count().T
```
