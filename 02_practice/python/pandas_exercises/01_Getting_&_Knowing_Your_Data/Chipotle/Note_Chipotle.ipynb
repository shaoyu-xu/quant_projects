# 1.shape attribute
## DataFrame: returns a tuple (number_of_rows, number_of_columns)
## Series: returns a tuple(number_of_elements,) notice the comma, it's a one-element tuple

```python
import pandas as pd
import numpy as np
```

```python
df = pd.DataFrame({'A': [1, 2, 3], 'B': [4, 5, 6]})
df.shape
```

```python
s = df['A']
s.shape
```

## Common pitfalls:
### No parentheses: df.shape is correct, df.shape() raises an error.
### Series ashape is a single-element tuple--you can't do s.shape[1]

# 3.info()
## Prints a concise summary of a DataFrame: number of rows, columns, column names, non-null counts, and data types(dtype). It does not return a value--it prints directly.

```python
df.info()
```

## Common use: Immediately check for missing values and verify data types after reading a CSV. If a numeric column shows object dtype, you may have messy data.

## Pitfall: info() is a DataFrame method--you can't call it on a Series.

## Excel analogy: Like clicking"Data" to "Get Data" or checking the "Used Range" properties: it gives you a bird's-eye view of the whole table.

# 4.groupby()

## Splits the data into groups based on one or more columns, then applies an aggregation function(mean(), sum(), std(), etc.) to each group.

```python
df.groupby('A')['B'].mean()
```

```python
df.groupby(['A', 'B'])['B'].last()
```

## Excel analogy: Exactly like a pivot table--you drag 'industry' to Rows and 'return' to Values, set to Average.

## Common financial use: Calculate average monthly return by industry; count how many stocks i each sector; group by data and ticker to get daily portfolio values.

## Key pitfall: groupby() alone returns a lazy GroupBy object. Nothing happens until you call an aggregation method. 

# 5.sort_values()

## Sorts a DataFrame of Series by the values in one or more columns.

```python
df.sort_values('return', ascending=False)      # highest return first
s.sort_values()                                 # sort a Series by its values
df.sort_values(['sector', 'pe'], ascending=[True, False])  # multi-column sort
```

## Excel analogy: The "Sort" dialog in Excel--pick a column, choose ascending or descending.

## Common financial use: Rank stocks by P/E ratio, find top 10 performers by monthly return, sort dates to ensure chronological order.

## Key pitfall:By default sort_values() returns a new object, leaving the original unchanged. Use inplace=True cautiously. 

# 6.numpy.round()

```python
import numpy as np
```

```python
arr = np.array([1.2345, 2.6789, 3.14159])
```

```python
np.round(arr, 2)
```

```python
np.round(arr)
```

```python
np.round(1234.567, -2)
```

## Key features:
  Works element-wise on arrays (vectorized)
  decimals parameter can be positive (decimal places), zero (integer), or negative (to the left of the decimal).
  For values exactly halfway (e.g., 2.5), it rounds to the nearest even number ("banker's rounding") to reduce statistical bias.
## Common pitfalls:
 This banker's rounding (round half to even) can be surprising: np.round(2.5) gives 2.0, while Python's built-in round(2.5) also gives 2 (Python 3 also uses banker's rounding). But np.round(3.5) gives 4.0. Know this when dealing with exact halves.

 For precise decimal rounding in finance (e.g., currency), consider Python's decimal module if you need strict decimal arithmetic (not floating-point).
## Financial use: Rounding stock prices to two decimals, formatting returns for display, rounding trade amounts.
